### Forword Interpolation 

```python
import math

# Predefined data points
x = [10, 20, 30, 40, 50]
y = [46, 66, 81, 93, 101]

n = len(x)

# Create forward difference table
diff = [[0 for j in range(n)] for i in range(n)]

for i in range(n):
    diff[i][0] = y[i]

for j in range(1, n):
    for i in range(n - j):
        diff[i][j] = diff[i + 1][j - 1] - diff[i][j - 1]

# Display the difference table
print("Forward Difference Table:\n")

for i in range(n):
    print(f"{x[i]:<8}", end="")
    for j in range(n - i):
        print(f"{diff[i][j]:<12.4f}", end="")
    print()

# Value to interpolate
xp = 25

# Newton Forward Interpolation
h = x[1] - x[0]
u = (xp - x[0]) / h

result = diff[0][0]
u_term = 1

for i in range(1, n):
    u_term *= (u - (i - 1))
    result += (u_term * diff[0][i]) / math.factorial(i)

print("\nValue to interpolate =", xp)
print("Interpolated value =", result)
```

```python
print("\nThe interpolated value at x =", xp, "is approximately", result)
```

### Backward Interpolation

```python
import math

# Predefined data points
x = [10, 20, 30, 40, 50]
y = [46, 66, 81, 93, 101]

n = len(x)

# Create backward/forward difference table
diff = [[0 for j in range(n)] for i in range(n)]

for i in range(n):
    diff[i][0] = y[i]

for j in range(1, n):
    for i in range(n - j):
        diff[i][j] = diff[i + 1][j - 1] - diff[i][j - 1]

# Display difference table
print("Difference Table:\n")

for i in range(n):
    print(f"{x[i]:<8}", end="")
    for j in range(n - i):
        print(f"{diff[i][j]:<12.4f}", end="")
    print()

# Value to interpolate
xp = 45

# Newton Backward Formula
h = x[1] - x[0]
u = (xp - x[-1]) / h

result = y[-1]
u_term = 1

for i in range(1, n):
    u_term *= (u + (i - 1))
    result += (u_term * diff[n - i - 1][i]) / math.factorial(i)

print("\nValue to interpolate =", xp)
print("Interpolated value =", result)
```

### Langrange Interpolation

```python
# Lagrange Interpolation

# Predefined data points
x = [10, 20, 30, 40, 50]
y = [46, 66, 81, 93, 101]

n = len(x)

# Value to interpolate
xp = 25

result = 0

# Lagrange Formula
for i in range(n):
    term = y[i]

    for j in range(n):
        if i != j:
            term *= (xp - x[j]) / (x[i] - x[j])

    result += term

print("Data Points:")
for i in range(n):
    print(f"x = {x[i]}, y = {y[i]}")

print("\nValue to interpolate =", xp)
print("Interpolated value =", result)
```

### Trapizal

```python
# Trapezoidal Rule using data points

x = [10, 20, 30, 40, 50]
y = [100, 400, 900, 1600, 2500  ]

h = x[1] - x[0]

result = (h / 2) * (y[0] + y[-1] + 2 * sum(y[1:-1]))

print("Data Points:")
for i in range(len(x)):
    print(f"x = {x[i]}, y = {y[i]}")

print("\nTrapezoidal Rule =", result)
```

### Simpson 1/3

```python
# Simpson's 1/3 Rule using data points

x = [10, 20, 30, 40, 50]
y = [100, 400, 900, 1600, 2500]

h = x[1] - x[0]

result = (h / 3) * (y[0] + 4 * sum(y[1:-1:2]) + 2 * sum(y[2:-1:2]) + y[-1]  )

print("Data Points:")
for i in range(len(x)):
    print(f"x = {x[i]}, y = {y[i]}")

print("\nSimpson's 1/3 Rule =", result)
```

### Simpson 3/8

```python
# Composite Simpson's 3/8 Rule using data points

# 7 data points (6 subintervals, which is a multiple of 3)
x = [0, 1, 2, 3, 4, 5, 6]
y = [1, 2, 5, 10, 17, 26, 37]   # y = x² + 1

n = len(x) - 1  # Number of subintervals

# Check if number of subintervals is a multiple of 3
if n % 3 != 0:
    print("Simpson's 3/8 Rule requires the number of subintervals to be a multiple of 3.")
else:
    h = x[1] - x[0]

    result = y[0] + y[-1]

    # Apply coefficients
    for i in range(1, n):
        if i % 3 == 0:
            result += 2 * y[i]      # Multiples of 3
        else:
            result += 3 * y[i]      # All other points

    result *= (3 * h) / 8

    print("Data Points:")
    for i in range(len(x)):
        print(f"x = {x[i]}, y = {y[i]}")

    print("\nComposite Simpson's 3/8 Rule =", result)
```

# Newton raphson

```python
# Newton-Raphson Method

def f(x):
    return x**3 - x - 2

def df(x):
    return 3*x**2 - 1

x = 1.5
tolerance = 0.0001
max_iterations = 20

print("Newton-Raphson Method")

for i in range(max_iterations):
    x_new = x - f(x) / df(x)

    print(f"Iteration {i+1}: x = {x_new:.6f}")

    if abs(x_new - x) < tolerance:
        break

    x = x_new

print("\nApproximate Root =", x_new)
```

### Regula Falsi Method

```python
# Regula Falsi Method

def f(x):
    return x**3 - x - 2

a = 1
b = 2
tolerance = 0.0001
max_iterations = 20

print("Regula Falsi Method")

if f(a) * f(b) >= 0:
    print("Invalid interval. f(a) and f(b) must have opposite signs.")
else:
    for i in range(max_iterations):
        c = (a * f(b) - b * f(a)) / (f(b) - f(a))

        print(f"Iteration {i+1}: c = {c:.6f}")

        if abs(f(c)) < tolerance:
            break

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    print("\nApproximate Root =", c)
```

###  Secant Method

```python
# Secant Method

def f(x):
    return x**3 - x - 2

x0 = 1
x1 = 2
tolerance = 0.0001
max_iterations = 20

print("Secant Method")

for i in range(max_iterations):
    if f(x1) - f(x0) == 0:
        print("Division by zero error.")
        break

    x2 = x1 - (f(x1) * (x1 - x0)) / (f(x1) - f(x0))

    print(f"Iteration {i+1}: x = {x2:.6f}")

    if abs(x2 - x1) < tolerance:
        break

    x0 = x1
    x1 = x2

print("\nApproximate Root =", x2)
```

### Bisection Method

```python
# Bisection Method

def f(x):
    return x**3 - x - 2

a = 1
b = 2
tolerance = 0.0001
max_iterations = 50

print("Bisection Method")

if f(a) * f(b) >= 0:
    print("Invalid interval. f(a) and f(b) must have opposite signs.")
else:
    for i in range(max_iterations):
        c = (a + b) / 2

        print(f"Iteration {i+1}: c = {c:.6f}")

        if abs(f(c)) < tolerance:
            break

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    print("\nApproximate Root =", c)
```

### Fixed Point Iteration 

```python
# Simple Iterative Method

def g(x):
    return (x + 2) ** (1 / 3)

x = 1.5
tolerance = 0.0001
max_iterations = 20

print("Simple Iterative Method")

for i in range(max_iterations):
    x_new = g(x)

    print(f"Iteration {i + 1}: x = {x_new:.6f}")

    if abs(x_new - x) < tolerance:
        break

    x = x_new

print("\nApproximate Root =", x_new)
```

### Jacobi 

```python
# Jacobi Method

x = 0
y = 0
z = 0

tolerance = 0.0001
max_iterations = 20

print("Jacobi Method")

for i in range(max_iterations):
    x_new = (12 - y - z) / 10
    y_new = (13 - 2*x - z) / 10
    z_new = (14 - 2*x - 2*y) / 10

    print(f"Iteration {i + 1}: x = {x_new:.6f}, y = {y_new:.6f}, z = {z_new:.6f}")

    if abs(x_new - x) < tolerance and abs(y_new - y) < tolerance and abs(z_new - z) < tolerance:
        break

    x = x_new
    y = y_new
    z = z_new

print("\nApproximate Solution:")
print("x =", x_new)
print("y =", y_new)
print("z =", z_new)
```

### Gauss-Seidel Method


```python
# Gauss-Seidel Method

x = 0
y = 0
z = 0

tolerance = 0.0001
max_iterations = 20

print("Gauss-Seidel Method")

for i in range(max_iterations):
    old_x = x
    old_y = y
    old_z = z

    x = (12 - y - z) / 10
    y = (13 - 2*x - z) / 10
    z = (14 - 2*x - 2*y) / 10

    print(f"Iteration {i + 1}: x = {x:.6f}, y = {y:.6f}, z = {z:.6f}")

    if abs(x - old_x) < tolerance and abs(y - old_y) < tolerance and abs(z - old_z) < tolerance:
        break

print("\nApproximate Solution:")
print("x =", x)
print("y =", y)
print("z =", z)
```

### Euler Method


```python

def f(x, y):
    return x + y

x = 0
y = 1

h = 0.1
xn = 0.5

print("Euler Method")
print(f"Initial: x = {x}, y = {y}")

while x < xn:
    y = y + h * f(x, y)
    x = x + h

    print(f"x = {x:.1f}, y = {y:.6f}")

print("\nApproximate y at x =", xn, "is", y)
```

### Runge-Kutta 4th Order Method (RK4)


```python
# Runge-Kutta 4th Order Method (RK4)

def f(x, y):
    return x + y

# Initial values
x = 0
y = 1

# Step size
h = 0.1

# Final value of x
xn = 0.5

print("Runge-Kutta 4th Order Method\n")
print(f"Initial: x = {x}, y = {y}\n")


while x < xn:
    k1 = h * f(x, y)
    k2 = h * f(x + h/2, y + k1/2)
    k3 = h * f(x + h/2, y + k2/2)
    k4 = h * f(x + h, y + k3)
    k = (k1 + 2*k2 + 2*k3 + k4) 
    y = y + k/ 6
    x = x + h

    print(f"x = {x:.1f}, y = {y:.6f}")

print("\nApproximate value of y =", y)
```
