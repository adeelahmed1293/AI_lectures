```python
a, b = 10, 3
a =10
b=3

print(a + b)    # 13  (addition)
print(a - b)    # 7   (subtraction)
print(a * b)    # 30  (multiplication)
print(a / b)    # 3.333  (division → float)
print(a // b)   # 3   (floor division)
print(a % b)    # 1   (modulo / remainder)
print(a ** b)   # 1000 (power: 10³)
```

```python

x=5

print(x == 5)   # True  (equal)
print(x != 3)   # True  (not equal)
print(x >  3)   # True  (greater)
print(x <  10)  # True  (less)
print(x >= 5)   # True  (greater or equal)
print(x <= 4)   # False (less or equal)
```

```python
a = True
b = False

print(a and b)   # False (both must be True)
print(a or  b)   # True  (at least one True)
print(not a)     # False (flips the value)

# Combined
x = 7
print(x > 5 and x < 10)  # True
```
