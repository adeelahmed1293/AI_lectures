```python
import numpy as np
import time

# Python list approach
lst = list(range(1_000_000))
t0 = time.time()
result = [x * 2 for x in lst]
print(f"List:  {(time.time()-t0)*1000:.1f} ms")

# NumPy approach
arr = np.arange(1_000_000)
t0 = time.time()
result = arr * 2
print(f"NumPy: {(time.time()-t0)*1000:.1f} ms")
```

```python
data = np.array([[2., 8., 5.],
                 [4., 6., 3.],
                 [6., 4., 7.]])
```

```python
mean = data.mean(axis=0)
```

```python
mean
```

```python
std  = data.std(axis=0)
```

```python
norm = (data - mean) / std   # broadcasts!

print(norm.mean(axis=0))    # ≈ [0, 0, 0]
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

C = A @ B          # preferred syntax (PEP 465)
D = np.dot(A, B)   # equivalent
print(C)
```

```python
# Solve: 3x + y = 9,  x + 2y = 8
A = np.array([[3., 1.],
              [1., 2.]])
b = np.array([9., 8.])

x = np.linalg.solve(A, b)
print(x)  # x = 2, y = 3
```
